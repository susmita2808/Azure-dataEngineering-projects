```python
#  Load Silver Stream Table

stream_df = spark.table("databricksrealtimehybrid.realtime_pipeline_silver.stream_orders")
```

```python
# KPI 1: Orders per City

from pyspark.sql.functions import count

orders_per_city_df = (
    stream_df
    .groupBy("city")
    .agg(count("order_id").alias("total_orders"))
)

display (orders_per_city_df)
```

```python
# KPI 2: Revenue per City

from pyspark.sql.functions import sum

revenue_per_city_df = (
    stream_df
    .groupBy("city")
    .agg(sum("amount").alias("total_revenue"))
)

display(revenue_per_city_df)
```

```python
# KPI 3: Real-Time Summary

summary_df = stream_df.agg(
    count("order_id").alias("total_orders"),
    sum("amount").alias("total_revenue")
)

display(summary_df)
```

**WRITE STREAM GOLD TABLES**

```python
# WRITE STREAM GOLD TABLES

orders_per_city_df.write.mode("overwrite").saveAsTable("databricksrealtimehybrid.realtime_pipeline_gold.orders_per_city")

revenue_per_city_df.write.mode("overwrite").saveAsTable("databricksrealtimehybrid.realtime_pipeline_gold.revenue_per_city")

summary_df.write.mode("overwrite").saveAsTable("databricksrealtimehybrid.realtime_pipeline_gold.realtime_summary")
```
